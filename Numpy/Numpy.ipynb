## 01 · Тип при создании

Сопоставьте float32, платформенный int и uint8 с dtype получившихся значений.

_Далее — исходные ячейки учебного материала._

```
import numpy as np
x = np.float32(1.0) 
x
```

```
y = np.int_([1,2,4]) # массив np
y            
```

```
z = np.arange(3, dtype=np.uint8) # np массив из 3х значений с с типом uint8
z
```

```
np.array([1, 2, 3], dtype='f')   # переводим массив в тип float32 
```

<a id="ai-035-section-02"></a>
## 02 · Преобразование типа

Посмотрите на astype, явные конструкторы и проверку семейства через issubdtype.

_Далее — исходные ячейки учебного материала._

```
z.astype(float)  # конвертируем uint8 в float
```

```
np.int8(z)       # конвертируем float в int8
```

```
z.dtype
```

```
d = np.dtype(int)
d
```

```
np.issubdtype(d, np.integer)  #проверка типа int
```

```
np.issubdtype(d, np.floating) #проверка типа float
```

## 03 · Диапазон и переполнение

Сравните степень с пределами iinfo до доверия напечатанному числу.

_Далее — исходные ячейки учебного материала._

```
np.power(100, 9, dtype=np.int64)
```

```
np.power(100, 8, dtype=np.int32)
```

```
np.iinfo(int)
```

```
np.iinfo(np.int32)
```

```
np.iinfo
```

```
np.power(100, 100, dtype=np.float64)
```
