## 06 · Подготовить checkpoint обучения

ModelCheckpoint только создаёт callback. Чтобы получить файл модели, нужны корректный путь, существующая папка, соответствующая метрика и fit с этим callback.

_Далее — исходные ячейки учебного материала._

#Создаем Keras Callback для сохранения нейронной сети на Google Drive

```
from tensorflow.keras.callbacks import ModelCheckpoint
filepath="/content/gdrive/'My Drive'/deep_nets/mynet_best.h5"
checkpoint = ModelCheckpoint(filepath, monitor='val_acc', verbose=1, 
                             save_best_only=True, mode='max')
```
