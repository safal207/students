<a id="ai-029-section-01"></a>
## 01 · Подготовка

Проверьте импорт библиотек и архивную команду выбора TensorFlow.

_Далее — исходные ячейки учебного материала._

```
%tensorflow_version 2.x
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Embedding, SimpleRNN 
from tensorflow.keras import utils
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline 
```

<a id="ai-029-section-02"></a>
## 02 · Словарь отзывов

Загрузка ограничивает словарь 10 000 индексами.

_Далее — исходные ячейки учебного материала._

## Загружаем данные

```
max_words=10000
```

```
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=max_words)
```

<a id="ai-029-section-03"></a>
## 03 · Подготовка последовательности

Padding задаёт единый вход из 200 позиций.

_Далее — исходные ячейки учебного материала._

## Подготовка данных для обучения

```
maxlen = 200
```

```
x_train = pad_sequences(x_train, maxlen=maxlen)
x_test = pad_sequences(x_test, maxlen=maxlen)
```

```
x_train[5002]
```

<a id="ai-029-section-04"></a>
## 04 · Небольшая модель

Две координаты Embedding и восемь единиц SimpleRNN задают компактную базу.

_Далее — исходные ячейки учебного материала._

## Создание нейронной сети

```
model = Sequential()
model.add(Embedding(max_words, 2, input_length=maxlen))
model.add(SimpleRNN(8))
model.add(Dense(1, activation='sigmoid'))
```

```
model.compile(optimizer='rmsprop', 
              loss='binary_crossentropy', 
              metrics=['accuracy'])
```

<a id="ai-029-section-05"></a>
## 05 · Обучение и кривые

Сопоставьте обучающую и проверочную метрики при RMSprop.

_Далее — исходные ячейки учебного материала._

## Обучаем нейронную сеть

```
history = model.fit(x_train, 
                    y_train, 
                    epochs=15,
                    batch_size=128,
                    validation_split=0.1)
```

```
plt.plot(history.history['accuracy'], 
         label='Доля верных ответов на обучающем наборе')
plt.plot(history.history['val_accuracy'], 
         label='Доля верных ответов на проверочном наборе')
plt.xlabel('Эпоха обучения')
plt.ylabel('Доля верных ответов')
plt.legend()
plt.show()
```

<a id="ai-029-section-06"></a>
## 06 · Оценка и интерпретация

Отделите вызов evaluate от исторического текстового комментария о точности.

_Далее — исходные ячейки учебного материала._

## Проверяем работу сети на тестовом наборе данных

```
scores = model.evaluate(x_test, y_test, verbose=1)
```

#Вывод: точность 58 %. Также сеть долго обчалась. Возможно следует добавить эпох или использовать архитектуры с другими слоями: LSTM и GRU
