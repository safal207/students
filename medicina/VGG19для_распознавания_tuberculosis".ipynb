## Создаем генераторы изображений

```
# Каталог с данными для обучения
train_dir = 'base_dir/train_dir'
test_dir = 'base_dir/val_dir'
# Размерность тензора на основе изображения для входных данных в нейронную сеть
# backend Tensorflow, channels_last
input_shape = (image_size, image_size, 3)
```

<a id="ai-060-section-03"></a>
## 03 · Аугментации и входная шкала

Обучающие изображения аугментируются, а проверочные только масштабируются. Нужны одинаковая предобработка и разделение по пациентам для сравнения двух вариантов VGG19.

_Далее — исходные ячейки учебного материала._

```
train_datagen = ImageDataGenerator(rescale=1. / 255,
    rotation_range=10,
    width_shift_range=0.1,
    height_shift_range=0.1,
    zoom_range=0.1,
    fill_mode='nearest')
```

Проверяем генератор

```
img_path = 'base_dir/train_dir/Normal/CHNCXR_0001_0.png'
img = image.load_img(img_path, target_size=(image_size, image_size))
x = image.img_to_array(img)
x = x.reshape((1,) + x.shape)
i = 0
for batch in train_datagen.flow(x, batch_size=1):
    plt.figure(i)
    imgplot = plt.imshow(image.array_to_img(batch[0]))
    i += 1
    if i % 6 == 0:
        break
plt.show()
```

Генератор для обучения

```
train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=(image_size, image_size),
    batch_size=batch_size,
    class_mode='binary')
```

Генератор для тестирования без дополнения данных

```
test_datagen = ImageDataGenerator(rescale=1. / 255)
```

```
test_generator = test_datagen.flow_from_directory(
    test_dir,
    target_size=(image_size, image_size),
    batch_size=batch_size,
    class_mode='binary')
```

Номера классов в генераторах

```
train_generator.class_indices
```

## Загружаем предварительно обученную нейронную сеть VGG19

<a id="ai-060-section-04"></a>
## 04 · Та же архитектура, другой режим

Здесь используется VGG19 с batch 32 и той же бинарной головой. Основные отличия от другого ноутбука относятся к параметрам обучения, а не названию сети.

_Далее — исходные ячейки учебного материала._

```
pretrained_model = VGG19(weights='imagenet', include_top=False)
```

vgg-ispravljeno-.png

```
pretrained_model.summary()
```

"Замораживаем" веса предварительно обученной нейронной сети

```
pretrained_model.trainable = False
```

## Создаем составную нейронную сеть

```
x = pretrained_model.output
x = GlobalAveragePooling2D()(x)
x = Dense(512)(x)
x = BatchNormalization()(x)
x = Activation('relu')(x)
x = Dropout(0.5)(x)
predictions = Dense(1, activation='sigmoid')(x)
model = Model(inputs=pretrained_model.input, outputs=predictions)
```

## Обучаем составную нейронную сеть на основе VGG19

```
model.compile(loss='binary_crossentropy',
              optimizer=RMSprop(lr=1e-4), 
              metrics=['accuracy'])
```

```
filepath="VGG19_best.h5"
checkpoint = ModelCheckpoint(filepath, monitor='val_accuracy', verbose=1, 
                             save_best_only=True, mode='max')
learning_rate_reduction = ReduceLROnPlateau(monitor='val_accuracy', 
                                            patience=3, 
                                            verbose=1, 
                                            factor=0.5, 
                                            min_lr=0.00001)
callbacks_list = [checkpoint, learning_rate_reduction]
```

<a id="ai-060-section-05"></a>
## 05 · Первый этап и контрольная точка

Первый этап: 23 эпохи, 40 обучающих и 30 проверочных шагов. Зафиксируйте фактический объём данных, чтобы сравнение с batch 23 было понятным.

_Далее — исходные ячейки учебного материала._

```
steps_per_epoch = 40
history = model.fit_generator(
    train_generator,
    steps_per_epoch=steps_per_epoch,
    validation_data=test_generator,
    validation_steps=30,
    callbacks=callbacks_list,
    epochs=23,
    verbose=2)
```

## 06 · Проверяем дообучаемые слои

После VGG19_best.h5 снова применяется граница 249; второй этап задан на 50 эпох по 100 шагов. Сначала проверьте наличие обучаемых слоёв.

_Далее — исходные ячейки учебного материала._

```
model.load_weights("VGG19_best.h5")
```

Размораживаем два последних блока vGG19

```
pretrained_model.trainable = False
for layer in model.layers[:249]:
   layer.trainable = False
for layer in model.layers[249:]:
   layer.trainable = True
```

Компилируем сеть с размороженными слоями

```
model.compile(loss='binary_crossentropy',
              optimizer=RMSprop(lr=1e-4), 
              metrics=['accuracy'])
```

```
filepath="VGG19_fine_tuned.h5"
checkpoint = ModelCheckpoint(filepath, monitor='val_accuracy', verbose=1, save_best_only=True, mode='max')
callbacks_list = [checkpoint, learning_rate_reduction]
```

```
steps_per_epoch = 100
history = model.fit_generator(
    train_generator,
    steps_per_epoch=steps_per_epoch,
    validation_data=test_generator,
    validation_steps=15,
    callbacks=callbacks_list,
    epochs=50,
    verbose=2)
```
