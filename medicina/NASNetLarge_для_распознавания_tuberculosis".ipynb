## 01 · Архитектура и учебный контекст

Код использует NASNetLarge. Это исторический опыт классификации изображений из набора с метками Normal/Tuberculosis, а не инструмент постановки диагноза.

_Далее — исходные ячейки учебного материала._

```
from tensorflow.keras.applications.nasnet import NASNetLarge
from tensorflow.keras.preprocessing import image
from tensorflow.keras.models import Model, load_model
from tensorflow.keras.callbacks import ModelCheckpoint, ReduceLROnPlateau
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, Dropout
from tensorflow.keras.layers import BatchNormalization, Activation
from tensorflow.keras.optimizers import Adam, RMSprop

from tensorflow.keras.preprocessing.image import ImageDataGenerator
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline 
```

```
# Размер изображения
image_size = 299
# Размер мини-выборки
batch_size = 32
```

## Создаем генераторы изображений

```
# Каталог с данными для обучения
train_dir = 'base_dir/train_dir'
test_dir = 'base_dir/val_dir'
# Размерность тензора на основе изображения для входных данных в нейронную сеть
# backend Tensorflow, channels_last
input_shape = (image_size, image_size, 3)
```

Генератор с дополнением данных (data augmentation)

<a id="ai-059-section-03"></a>
## 03 · Аугментации и входная шкала

Генераторы создают RGB-тензоры 299 × 299 с делением на 255. До восстановления опыта проверьте соответствие размера и шкалы выбранным NASNetLarge-весам.

_Далее — исходные ячейки учебного материала._

```
train_datagen = ImageDataGenerator(rescale=1. / 255,
    rotation_range=10,
    width_shift_range=0.1,
    height_shift_range=0.1,
    zoom_range=0.1,
    fill_mode='nearest')
```

Проверяем генератор

```
img_path = 'base_dir/train_dir/Normal/CHNCXR_0001_0.png'
img = image.load_img(img_path, target_size=(image_size, image_size))
x = image.img_to_array(img)
x = x.reshape((1,) + x.shape)
i = 0
for batch in train_datagen.flow(x, batch_size=1):
    plt.figure(i)
    imgplot = plt.imshow(image.array_to_img(batch[0]))
    i += 1
    if i % 6 == 0:
        break
plt.show()
```

Генератор для обучения

```
train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=(image_size, image_size),
    batch_size=batch_size,
    class_mode='binary')
```

Генератор для тестирования без дополнения данных

```
test_datagen = ImageDataGenerator(rescale=1. / 255)
```

```
test_generator = test_datagen.flow_from_directory(
    test_dir,
    target_size=(image_size, image_size),
    batch_size=batch_size,
    class_mode='binary')
```

Номера классов в генераторах

```
train_generator.class_indices
```

## Загружаем предварительно обученную нейронную сеть Inception

<a id="ai-059-section-04"></a>
## 04 · Большая предобученная основа

Фактически создаётся NASNetLarge, несмотря на старые подписи Inception. Новая голова остаётся бинарной, поэтому архитектурную сложность нельзя путать с количеством классов.

_Далее — исходные ячейки учебного материала._

```
pretrained_model = NASNetLarge(weights='imagenet', include_top=False)
```

```
pretrained_model.summary()
```

"Замораживаем" веса предварительно обученной нейронной сети

```
pretrained_model.trainable = False
```

## Создаем составную нейронную сеть

```
x = pretrained_model.output
x = GlobalAveragePooling2D()(x)
x = Dense(512)(x)
x = BatchNormalization()(x)
x = Activation('relu')(x)
x = Dropout(0.5)(x)
predictions = Dense(1, activation='sigmoid')(x)
model = Model(inputs=pretrained_model.input, outputs=predictions)
```

## Обучаем составную нейронную сеть на основе NASNetLarge


```
model.compile(loss='binary_crossentropy',
              optimizer=RMSprop(lr=1e-4), 
              metrics=['accuracy'])
```

```
filepath="NASNetLarge_best.h5"
checkpoint = ModelCheckpoint(filepath, monitor='val_accuracy', verbose=1, 
                             save_best_only=True, mode='max')
learning_rate_reduction = ReduceLROnPlateau(monitor='val_accuracy', 
                                            patience=3, 
                                            verbose=1, 
                                            factor=0.5, 
                                            min_lr=0.00001)
callbacks_list = [checkpoint, learning_rate_reduction]
```

<a id="ai-059-section-05"></a>
## 05 · Первый этап и контрольная точка

Первый цикл содержит 125 эпох по 24 обучающих шага и 20 проверочных. Сверьте эти числа с длинами генераторов и доступной памятью.

_Далее — исходные ячейки учебного материала._

```
steps_per_epoch = 24
history = model.fit_generator(
    train_generator,
    steps_per_epoch=steps_per_epoch,
    validation_data=test_generator,
    validation_steps=20,
    callbacks=callbacks_list,
    epochs=125,
    verbose=2)
```

## Дообучение сверточных слоев NASNetLarge

Загружаем лучшие веса

<a id="ai-059-section-06"></a>
## 06 · Проверяем дообучаемые слои

Второй этап использует границу 249 и всего 5 эпох по 10 шагов. Требуется проверить состав trainable-слоёв и реальный объём обновлений.

_Далее — исходные ячейки учебного материала._

```
model.load_weights("NASNetLarge_best.h5")
```

Размораживаем два последних блока Inception

```
pretrained_model.trainable = False
for layer in model.layers[:249]:
   layer.trainable = False
for layer in model.layers[249:]:
   layer.trainable = True
```

Компилируем сеть с размороженными слоями

```
model.compile(loss='binary_crossentropy',
              optimizer=RMSprop(lr=1e-4), 
              metrics=['accuracy'])
```

```
filepath="NASNetLarge_fine_tuned.h5"
checkpoint = ModelCheckpoint(filepath, monitor='val_accuracy', verbose=1, save_best_only=True, mode='max')
callbacks_list = [checkpoint, learning_rate_reduction]
```

```
steps_per_epoch = 10
history = model.fit_generator(
    train_generator,
    steps_per_epoch=steps_per_epoch,
    validation_data=test_generator,
    validation_steps=5,
    callbacks=callbacks_list,
    epochs=5,
    verbose=2)
```
